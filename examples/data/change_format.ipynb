```python
import numpy as np
import torch
from sklearn.preprocessing import LabelEncoder
```

```python
# Cargar el archivo numpy
file_path = "D:\\0_CUCEI\\Modular\\conn2res_torch\\examples\\data\\human\\rsn_mapping.npy"
data = np.load(file_path, allow_pickle=True)
```

```python
# 1. Análisis inicial del archivo
print("=== Análisis del archivo ===")
print("Tipo de objeto:", type(data))
print("Forma del array:", data.shape)
print("Tipo de datos:", data.dtype)
```

```python
# 2. Inspección de contenido
print("\n=== Muestra del contenido ===")
if data.ndim == 1:
    print("Primeros 5 elementos:", data[:5])
elif data.ndim == 2:
    print("Primeras 5 filas:\n", data[:5])
else:
    print("Array multidimensional. Primero elemento:", data[0])
```

```python
# 3. Detección de strings
def contains_strings(array):
    if array.dtype == object:
        return any(isinstance(x, str) for x in array.flat)
    return False

if contains_strings(data):
    print("\nEl array contiene strings. Aplicando encoding...")
    
    # Opción 1: Label Encoding para categorías
    if data.ndim == 1:
        le = LabelEncoder()
        data_encoded = le.fit_transform(data)
        print("\nMapping de categorías:")
        for i, cat in enumerate(le.classes_):
            print(f"{cat} -> {i}")
    else:
        print("Array multidimensional con strings. Requiere procesamiento especial.")
        # Aquí necesitarías expandir según tu estructura específica
else:
    print("\nNo se detectaron strings. Convirtiendo directamente a tensor.")
    data_encoded = data
```

```python
# 4. Conversión a tensor PyTorch
try:
    tensor_data = torch.from_numpy(data_encoded if 'data_encoded' in locals() else data)
    print("\n=== Tensor resultante ===")
    print("Forma:", tensor_data.shape)
    print("Tipo de datos:", tensor_data.dtype)
    
    # 5. Guardar el tensor procesado (opcional)
    output_path = file_path.replace('.npy', '_processed.pt')
    torch.save(tensor_data, output_path)
    print(f"\nTensor guardado en: {output_path}")
    
except Exception as e:
    print("\nError en la conversión:", str(e))
    print("Recomendación: Verifica la estructura de datos y aplica preprocesamiento adecuado.")
```

# Divisor

```python
# 1. Análisis de categorías únicas
unique_labels, counts = np.unique(data, return_counts=True)
print("Categorías encontradas y su frecuencia:")
for label, count in zip(unique_labels, counts):
    print(f"{label}: {count} elementos")
```

```python
# 2. Aplicar Label Encoding
le = LabelEncoder()
data_encoded = le.fit_transform(data)
```

```python
# 3. Convertir a tensor PyTorch
tensor_data = torch.from_numpy(data_encoded).long()  # Usamos long para etiquetas

print("\nResultado de la conversión:")
print(f"Tensor shape: {tensor_data.shape}")
print(f"Tensor dtype: {tensor_data.dtype}")
print(f"Valores únicos en tensor: {torch.unique(tensor_data)}")
```

```python
# 4. Opcional: Guardar el mapeo de categorías
label_mapping = {i: label for i, label in enumerate(le.classes_)}
print("\nMapeo de categorías a índices:")
print(label_mapping)
```

```python
# 5. Guardar el tensor procesado
output_path = file_path.replace('.npy', '_processed.pt')
torch.save(tensor_data, output_path)
print(f"\nTensor guardado en: {output_path}")
```
